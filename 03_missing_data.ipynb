# Handling Missing Data

Explore how to detect, count, remove, and fill missing values in a pandas DataFrame. The examples use `NaN` to represent missing data.

**Topics:** `isna`, `dropna`, and `fillna`.


## Inspect missing values


```python
import pandas as pd  
import numpy as np
```

## Detecting missing values


```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
df.isna()  # True marks a missing value.
```

```python
# Count missing values in each column. 
df.isna().sum()
```

```python
# Check whether each column contains at least one missing value. 
df.isna().any()
```

REMOVING MISSING DATA  #work with column wise 

```python
df.dropna()  # Return rows that contain no missing values. 
```

```python
df.dropna(thresh=3)  # Keep rows with at least 3 non-missing values. 
```

## Filling the missing data #column wise


```python
df.fillna(0)  # Return a copy with missing values replaced by 0.
```

```python
values={'A':0,'B':100,'C':300,'D':400}
df.fillna(value=values)
```

```python
# Fill numeric missing values with each column's mean. 
df.fillna(df.mean())
```
