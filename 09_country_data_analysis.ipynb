```python
# Q8: Which country has the second-largest population, and who is its political leader?
```

```python
df.nlargest(2, "population").iloc[[1]][["country", "political_leader"]]  # Second-largest population.
```

```python
# Q9: Which countries have a missing political-leader value? 
```

```python
df.loc[df["political_leader"].isna(), "country"]  # Countries with unknown leader values.
```

```python
# Q10: How many full country names contain the word "republic"? 
```

```python
# Count country full names containing the word "republic" (case-insensitive).
republic_count = df["country_long"].str.contains("republic", case=False, na=False).sum()
republic_count    
```

```python
# Q11: Which country in Africa has the largest population?
```

```python
africa_df = df[df["continent"] == "Africa"]
africa_df.loc[africa_df["population"].idxmax(), "country"]
```
