# Merging, Concatenating, and Joining DataFrames

Compare common ways to combine DataFrames: database-style merges, stacking rows or columns with `concat`, and index-based joins.

**Topics:** inner, outer, left, and right merges; concatenation; index joins.


## Merging joining concatination


```python
import numpy as  np 
import pandas as pd
```

## Merge two DataFrames


```python
employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# DataFrame 2: Salary information
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})
```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees, salaries)  # Defaults to an inner merge on shared column names. 
```

```python
pd.merge(employees, salaries, on="employee_id", how="inner")  # Keep IDs present in both tables.
```

```python
pd.merge(employees, salaries, on="employee_id", how="outer")  # Keep every ID from both tables.
```

```python
pd.merge(employees, salaries, on="employee_id", how="left")  # Keep every employee.
```

```python
pd.merge(employees, salaries, on="employee_id", how="right")  # Keep every salary record.
```

## Concatenate DataFrames


```python
import pandas as pd 
import numpy as np 
```

```python
df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2'],
    'C': ['C0', 'C1', 'C2']
})

df2 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5'],
    'C': ['C3', 'C4', 'C5']
})
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1, df2], axis=0, ignore_index=True)  # Stack rows vertically. 
```

```python
pd.concat([df1, df2], axis=1)  # Place columns side by side, aligning index labels. 
```
