# Pandas DataFrames

Practice creating DataFrames from dictionaries and lists, selecting columns and rows, adding and removing columns, and filtering rows with conditions.

**Topics:** construction, indexing, selection, and conditional filtering.


## Data frames


```python
#Creating a data frame 
```

```python
import numpy as np
import pandas as pd 
```

```python
data={
    'name': ['John', 'Anna', 'Peter', 'Linda'],
    'age':[28,34,29,42],
    'city': ['New York', 'Paris', 'Berlin', 'London'],
    'salary':[65000,70000,62000,85000]
}
```

```python
df=pd.DataFrame(data)
df
```

```python
data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]
df2=pd.DataFrame(data_list)
df2
```

```python
#if i want to give column name instead of indexes 
columns=['name','age','city','salary']
df2=pd.DataFrame(data_list,columns=columns)
df2
```

## Selecting and indexing columns


```python
#if i want entire name column 
df2["name"]
```

```python
# Select more than one column by passing a list of column names. 
df2[["name","city"]]
```

```python
#Creating a new column 
df2["designation"]=["doctor","engineer","doctor","engineer"]
df2
```

```python
#removing column
```

```python
# This raises a KeyError because drop() defaults to dropping rows.
# To remove a column, specify columns=[...] or axis=1.
# df2.drop("designation") 
```

```python
df2.drop(columns=["designation"])
```

```python
# drop() returns a new DataFrame by default; df2 is unchanged unless reassigned or inplace=True.
df2
```

```python
# Remove the column in place for this example.
df2.drop(columns=["designation"], inplace=True)
df2
```

```python
#for deleting row 
df2.drop(index=0, inplace=True)
df2
```

```python
#selecting rows   we us loc[index value ]
```

```python
df2.loc[1]
```

```python
#multiple rows
df2.loc[[1,2]] # index 1 and 2 
```

## Selecting subsets of rows and columns


```python
df2.loc[[1, 2]]
```

```python
df2.loc[[1, 2]][["city","salary"]]
```

## Conditional selection


```python
# A Boolean condition can be used to filter rows.

df2["age"]>30
```

```python
df2[df2["age"]>30]
```

```python
#only want people whose age is above 30 and their city must be paris 
df2[(df2["age"]>30) & (df2["city"]=="Paris")]
```
