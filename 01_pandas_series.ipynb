# Pandas Series

Learn how to create a pandas `Series` from a list, NumPy array, or dictionary, and how to assign custom index labels.

**Topics:** Series creation, values, and custom indexes.


## Pandas series  #creating


```python
import numpy as np
import pandas as pd
```

```python
labels=['a','b','c']
my_list=[10,20,30]
arr=np.array([10,20,30])
d={'a':10,'b':20,'c':30}
```

```python
pd.Series(my_list)
```

```python
# Assign custom labels as the Series index. 
pd.Series(my_list, index=labels)
```

## Create a Series from other Python objects


```python
pd.Series(arr, index=labels)  # Create a Series from a NumPy array.
```

```python
pd.Series(d)  # A dictionary uses its keys as the index labels.
```
