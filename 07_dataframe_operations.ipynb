## Data frame basic operation


```python
df1 = pd.DataFrame({
    'A': [1, 2, 3, 4, 5],
    'B': [10, 20, 30, 40, 50],
    'C': [100, 200, 300, 400, 500]
})
```

## Data frame applying function


```python
def square(x):
    return x**2
```

```python
df1["A_squared"] = df1["A"].apply(square)
df1  # Add a new column containing squared values.
```
