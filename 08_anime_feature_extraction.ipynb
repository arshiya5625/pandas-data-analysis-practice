```python
def extract_episode_count(title):
    """Extract the episode count from title text such as '(64 eps)'."""
    match = re.search(r"\((\d+)\s*eps\)", str(title), flags=re.IGNORECASE)
    return int(match.group(1)) if match else pd.NA    
            
```

```python
df["Episodes"] = df["Title"].apply(extract_episode_count).astype("Int64")
```

```python
def extract_airing_period(title):
    """Extract the airing date text that follows the episode count."""
    match = re.search(r"\)\s*(.*?)(?:\d[\d,]* members|Manga Store|$)", str(title))
    return match.group(1).strip() if match else pd.NA    
```

```python
df["Airing Period"] = df["Title"].apply(extract_airing_period)
```
